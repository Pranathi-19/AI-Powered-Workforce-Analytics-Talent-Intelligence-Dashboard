##Hospital Management System (SQL)

Create a Hospital Management System using SQL. Design the database, insert sample data, and write queries to retrieve, update, and delete records. Include joins, aggregate functions, grouping, sorting, and filtering. The queries should be easy to understand and run without errors in google colab

**1: Import SQLite and Create Database**

```python
import sqlite3

# Create database
conn = sqlite3.connect("hospital.db")
cursor = conn.cursor()

print("Hospital Database Created Successfully")
```

**2: Create Tables**

```python
cursor.executescript("""
CREATE TABLE Patients(
    PatientID INTEGER PRIMARY KEY,
    PatientName TEXT,
    Age INTEGER,
    Gender TEXT,
    City TEXT
);

CREATE TABLE Doctors(
    DoctorID INTEGER PRIMARY KEY,
    DoctorName TEXT,
    Specialization TEXT,
    Experience INTEGER
);

CREATE TABLE Appointments(
    AppointmentID INTEGER PRIMARY KEY,
    PatientID INTEGER,
    DoctorID INTEGER,
    AppointmentDate TEXT,
    ConsultationFee REAL,
    FOREIGN KEY(PatientID) REFERENCES Patients(PatientID),
    FOREIGN KEY(DoctorID) REFERENCES Doctors(DoctorID)
);
""")

conn.commit()

print("Tables Created Successfully")
```

**3: Insert Sample Data**

```python
cursor.executescript("""

INSERT INTO Patients VALUES
(101,'Aarav',25,'Male','Hyderabad'),
(102,'Diya',30,'Female','Warangal'),
(103,'Kiran',42,'Male','Vijayawada'),
(104,'Sneha',27,'Female','Hyderabad'),
(105,'Rahul',35,'Male','Karimnagar');

INSERT INTO Doctors VALUES
(201,'Dr. Meera','Cardiology',12),
(202,'Dr. Rakesh','Orthopedics',8),
(203,'Dr. Kavya','Dermatology',6),
(204,'Dr. Arjun','Neurology',15);

INSERT INTO Appointments VALUES
(301,101,201,'2026-07-01',700),
(302,102,203,'2026-07-02',500),
(303,103,202,'2026-07-03',650),
(304,104,204,'2026-07-04',900),
(305,105,201,'2026-07-05',700);

""")

conn.commit()

print("Sample Data Inserted")
```

**4: View All Patients**

```python
cursor.execute("SELECT * FROM Patients")

for row in cursor.fetchall():
    print(row)
```

**5: WHERE Clause (Filtering)**

```python
cursor.execute("""
SELECT * FROM Patients
WHERE City='Hyderabad'
""")

for row in cursor.fetchall():
    print(row)
```

**6: ORDER BY (Sorting)**

```python
cursor.execute("""
SELECT * FROM Doctors
ORDER BY Experience DESC
""")

for row in cursor.fetchall():
    print(row)
```

**7: Aggregate Functions**

```python
#COUNT:
cursor.execute("""
SELECT COUNT(*) FROM Patients
""")

print(cursor.fetchone())

#AVG:
cursor.execute("""
SELECT AVG(ConsultationFee)
FROM Appointments
""")

print(cursor.fetchone())

#MAX:
cursor.execute("""
SELECT MAX(ConsultationFee)
FROM Appointments
""")

print(cursor.fetchone())

#MIN:
cursor.execute("""
SELECT MIN(ConsultationFee)
FROM Appointments
""")

print(cursor.fetchone())

#SUM:
cursor.execute("""
SELECT SUM(ConsultationFee)
FROM Appointments
""")

print(cursor.fetchone())
```

**8: GROUP BY**

```python
cursor.execute("""

SELECT City,
COUNT(*) AS TotalPatients

FROM Patients

GROUP BY City

""")

for row in cursor.fetchall():
    print(row)
```

**9: INNER JOIN**

```python
cursor.execute("""

SELECT
Patients.PatientName,
Doctors.DoctorName,
Doctors.Specialization,
Appointments.ConsultationFee

FROM Appointments

INNER JOIN Patients
ON Patients.PatientID=Appointments.PatientID

INNER JOIN Doctors
ON Doctors.DoctorID=Appointments.DoctorID

""")

for row in cursor.fetchall():
    print(row)
```

**10: UPDATE Record**

```python
cursor.execute("""

UPDATE Patients

SET City='Secunderabad'

WHERE PatientID=105

""")

conn.commit()

print("Record Updated")

#Verify:
cursor.execute("SELECT * FROM Patients WHERE PatientID=105")
print(cursor.fetchone())
```

**11: DELETE Record**

```python
cursor.execute("""

DELETE FROM Patients

WHERE PatientID=104

""")

conn.commit()

print("Record Deleted")

#Verify:
cursor.execute("SELECT * FROM Patients")

for row in cursor.fetchall():
    print(row)
```

**12: Close Database**

```python
conn.close()

print("Database Closed Successfully")
```
