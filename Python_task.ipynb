**Online Food Ordering System**

1: Variables and Data Types

```python
restaurant_name = "Spice Garden"

food_id = 501
food_name = "Veg Biryani"
category = "Main Course"

price = 180.50
delivery_charge = 40.0

available = True
veg = True

print("Restaurant:", restaurant_name)
print("Food ID:", food_id)
print("Food Name:", food_name)
print("Category:", category)
print("Price: ₹", price)
print("Delivery Charge: ₹", delivery_charge)
print("Available:", available)
print("Vegetarian:", veg)
```

2: Input and Output

```python
customer = input("Enter Customer Name: ")
order_item = input("Enter Food Item: ")
quantity = int(input("Enter Quantity: "))

print("\nOrder Summary")
print("Customer:", customer)
print("Food Item:", order_item)
print("Quantity:", quantity)
```

3: Operators

```python
food_price = 180.50
delivery_charge = 40

subtotal = food_price * quantity
grand_total = subtotal + delivery_charge

print("Subtotal = ₹", subtotal)
print("Delivery Charge = ₹", delivery_charge)
print("Grand Total = ₹", grand_total)
```

4: Conditional Statements

```python
if quantity == 1:
    print("Single Item Order")
elif quantity <= 5:
    print("Regular Order")
else:
    print("Bulk Order")
```

5: Food Availability

```python
available = True

if available:
    print("Food is available.")
else:
    print("Food is currently unavailable.")
```

6: Delivery Eligibility

```python
distance = float(input("Enter Delivery Distance (km): "))

if distance <= 5:
    print("Free Delivery")
elif distance <= 10:
    print("Standard Delivery Charge")
else:
    print("Delivery Not Available")
```

7: Discount Calculation

```python
if grand_total >= 500:
    discount = grand_total * 0.10
    final_amount = grand_total - discount
    print("Discount = ₹", discount)
    print("Amount to Pay = ₹", final_amount)
else:
    print("No Discount")
    print("Amount to Pay = ₹", grand_total)
```

8: Payment Method

```python
payment = input("Enter Payment Method (Cash/UPI/Card): ")

if payment.lower() == "cash":
    print("Cash on Delivery Selected")
elif payment.lower() == "upi":
    print("UPI Payment Successful")
elif payment.lower() == "card":
    print("Card Payment Successful")
else:
    print("Invalid Payment Method")
```

9: Final Bill

```python
print("\n========== FINAL BILL ==========")
print("Customer Name :", customer)
print("Food Ordered  :", order_item)
print("Quantity      :", quantity)
print("Bill Amount   : ₹", grand_total)
print("Payment Mode  :", payment)
print("--------------------------------")
print("Thank you for ordering from", restaurant_name)
print("Have a Great Day!")
```
